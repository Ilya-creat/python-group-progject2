```python
#параметры группировки лиг
international_tournaments = [
    'UEFA Champions League', 'UEFA Europa League', 'UEFA Nations League', 'Copa America', 'CONMEBOL Sudamericana',
    'CAF Champions League', 'CONMEBOL Libertadores', 'FIFA Club World Cup', 'CONCACAF Champions League', 'AFC Champions League',
    'AFC Cup', 'African Nations Championship', 'CAF Confederation Cup', 'CONCACAF Gold Cup', 'EAFF E-1 Football Championship',
    'World Cup - Qualification Europe', 'Asian Cup - Qualification', 'UEFA U21 Championship', 'UEFA Youth League',
    'Leagues Cup', 'MLS All-Star', 'Campeones Cup', 'ASEAN Championship U23', 'Tournoi Maurice Revello',
    'UEFA U19 Championship', 'COSAFA Cup', 'CAFA Nations Cup', 'Concacaf Central American Cup',
    'Premier League International Cup', 'UEFA Nations League - Women', 'UEFA Champions League Women', 'UEFA Super Cup',
    'Copa America Femenina', 'AFC Challenge League', "AFC Women's Champions League", 'AGCFF Gulf Champions League',
    'FIFA Intercontinental Cup', 'EAFF E-1 Football Championship - Women', 'UEFA Europa Cup - Women',
    'CONMEBOL Nations League Women', 'CONCACAF Caribbean Club Championship', 'CONCACAF W Champions Cup', 'CONCACAF Women U20',
]

women_keywords = ['Women', 'Femenina', 'Frauen', 'WSL', 'NWSL', 'Kvindeliga', 'WK-League']
cup_keywords    = ['Cup', 'Trofeo', 'Pokal', 'Beker', 'Taça', 'Shield', 'Supercupa', 'Copa']

if 'bookmaker_margin' not in df.columns:
    for col in ['Match Winner_Home', 'Match Winner_Draw', 'Match Winner_Away']:
        df[col] = pd.to_numeric(df[col], errors='coerce')
    df['bookmaker_margin'] = (1/df['Match Winner_Home']
                              + 1/df['Match Winner_Draw']
                              + 1/df['Match Winner_Away'] - 1)

league_names = df['leagueName'].dropna().unique().tolist()

womens_leagues = [name for name in league_names if any(k in str(name) for k in women_keywords)]
cup_tournaments = [name for name in league_names if any(k in str(name) for k in cup_keywords)]

international_set = set(international_tournaments)
women_set         = set(womens_leagues) - international_set
cup_set           = set(cup_tournaments) - international_set - women_set
domestic_set      = set(league_names) - international_set - women_set - cup_set

def league_group(name):
    if name in international_set:
        return 'International'
    if name in women_set:
        return 'Women'
    if name in cup_set:
        return 'Cups'
    return 'Domestic'

df['league_group'] = df['leagueName'].map(league_group)
```

```python
def plot_box_by_group(ycol, title, ylim=None, top_n_per_group=None):
    """
    Рисует boxplot по каждой группе лиг (ось X — лиги внутри группы).
    Можно ограничить количество лиг top_n_per_group по числу матчей, чтобы графики были читабельнее.
    """
    for grp in ['International', 'Women', 'Cups', 'Domestic']:
        dfg = df[df['league_group'] == grp].copy()
        dfg[ycol] = pd.to_numeric(dfg[ycol], errors='coerce')

        if dfg.empty:
            continue

        order = (
            dfg.groupby('leagueName')[ycol].median()
            .sort_values()
            .index
            .tolist()
        )

        if top_n_per_group is not None:
            top_leagues = (
                df['leagueName'].value_counts()
                .head(top_n_per_group)
                .index
                .tolist()
            )
            order = [x for x in order if x in top_leagues]
            dfg = dfg[dfg['leagueName'].isin(order)]

        plt.figure(figsize=(14, 7))
        sns.boxplot(x='leagueName', y=ycol, data=dfg, order=order)
        if ylim is not None:
            plt.ylim(*ylim)
        plt.title(f'{title} — группа: {grp}')
        plt.xlabel('Лига')
        plt.ylabel(ycol)
        plt.xticks(rotation=60)
        plt.show()
```

```python
plot_box_by_group('Match Winner_Away',
                  title='Распределение коэффициента Match Winner_Away по лигам',
                  ylim=(0, 10), top_n_per_group=15)

plot_box_by_group('Match Winner_Draw',
                  title='Распределение коэффициента Match Winner_Draw по лигам',
                  ylim=(0, 10), top_n_per_group=15)

plot_box_by_group('Match Winner_Home',
                  title='Распределение коэффициента Match Winner_Home по лигам',
                  ylim=(0, 10), top_n_per_group=15)

plot_box_by_group('bookmaker_margin',
                  title='Распределение маржи букмекеров по лигам',
                  ylim=(0, 0.15), top_n_per_group=15)
```

```python
import numpy as np
import pandas as pd

# диапазоны коэффициентов
bins = np.arange(1.0, 10.5, 0.5)
labels = [f"[{round(b,1)}–{round(b+0.5,1)})" for b in bins[:-1]]

for col in ['Match Winner_Home', 'Match Winner_Draw', 'Match Winner_Away']:
    df[col] = pd.to_numeric(df[col], errors='coerce')

df['match_outcome'] = np.select(
    [
        df['homeResult'] > df['awayResult'],
        df['homeResult'] == df['awayResult'],
        df['homeResult'] < df['awayResult']
    ],
    ['Home', 'Draw', 'Away'],
    default=None
)

# анализ по типу исхода
def analyze_outcome(outcome_col, implied_prob_col, outcome_label):
    df_temp = df.copy()
    df_temp['odds_bin'] = pd.cut(df_temp[outcome_col], bins=bins, labels=labels, right=False)

    grouped = (df_temp.groupby('odds_bin').agg(mean_implied_prob=(implied_prob_col, 'mean'),actual_frequency=('match_outcome', lambda x: np.mean(x == outcome_label))).dropna())
    grouped['diff'] = grouped['actual_frequency'] - grouped['mean_implied_prob']
    return grouped

home_analysis = analyze_outcome('Match Winner_Home', 'implied_home_prob', 'Home')
draw_analysis = analyze_outcome('Match Winner_Draw', 'implied_draw_prob', 'Draw')
away_analysis = analyze_outcome('Match Winner_Away', 'implied_away_prob', 'Away')
```

```python
plt.figure(figsize=(10,5))
plt.plot(home_analysis.index, home_analysis['mean_implied_prob'], label='Подразумеваемая вероятность', marker='o')
plt.plot(home_analysis.index, home_analysis['actual_frequency'], label='Фактическая частота', marker='o')
plt.title('Победа хозяев: фактическая vs подразумеваемая вероятность')
plt.xlabel('Диапазон коэффициентов')
plt.ylabel('Вероятность')
plt.legend()
plt.xticks(rotation=45)
plt.show()
```

```python
# Рассчитываем абсолютное отклонение
home_analysis['abs_diff'] = home_analysis['diff'].abs()

home_sorted = home_analysis.sort_values('abs_diff', ascending=False)

print("Диапазоны коэффициентов с наибольшей разницей между фактической и подразумеваемой вероятностью (победа хозяев):")
print(home_sorted[['mean_implied_prob', 'actual_frequency', 'diff']].head(10))
```

```python
def plot_implied_vs_actual(df_analysis, outcome_label):
    """
    df_analysis — таблица с колонками mean_implied_prob и actual_frequency
    outcome_label — подпись графика ("Победа хозяев", "Ничья", "Победа гостей")
    """
    plt.figure(figsize=(10,5))
    plt.plot(df_analysis.index, df_analysis['mean_implied_prob'],
             marker='o', label='Подразумеваемая вероятность (1/коэф.)')
    plt.plot(df_analysis.index, df_analysis['actual_frequency'],
             marker='o', label='Фактическая частота исхода')
    plt.title(f'{outcome_label}: фактическая частота vs подразумеваемая вероятность')
    plt.xlabel('Диапазон коэффициентов')
    plt.ylabel('Вероятность')
    plt.xticks(rotation=45)
    plt.legend()
    plt.grid(True, linestyle='--', alpha=0.5)
    plt.show()
```

```python
plot_implied_vs_actual(home_analysis, "🏠 Победа хозяев")
plot_implied_vs_actual(draw_analysis, "🤝 Ничья")
plot_implied_vs_actual(away_analysis, "🚗 Победа гостей")
```

```python
def plot_diff_bar(df_analysis, outcome_label):
    plt.figure(figsize=(10,5))
    sns.barplot(x=df_analysis.index, y=df_analysis['diff'], palette='coolwarm')
    plt.axhline(0, color='black', linestyle='--', linewidth=1)
    plt.title(f'{outcome_label}: разница между фактической и подразумеваемой вероятностью')
    plt.xlabel('Диапазон коэффициентов')
    plt.ylabel('Разница (actual - implied)')
    plt.xticks(rotation=45)
    plt.grid(axis='y', linestyle='--', alpha=0.5)
    plt.show()
```

```python
plot_diff_bar(home_analysis, 'Победа хозяев')
plot_diff_bar(draw_analysis, 'Ничья')
plot_diff_bar(away_analysis, 'Победа гостей')
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

def norm_country(s: str) -> str:
    if pd.isna(s):
        return ''
    s = str(s).strip().replace('_','-')
    s = '-'.join(s.split())
    return s

world_class_football = [
    'Germany','France','Spain','Italy','Great Britain','Portugal','Netherlands','Belgium','Switzerland','Sweden',
    'Brazil','Argentina','Chile','Uruguay','Colombia','Cameroon','Nigeria','Egypt','Ghana',
    'Japan','South-Korea','Iran','USA','Mexico','Canada','Australia'
]
strong_developing_leagues = [
    'Poland','Russia','Turkey','Czech-Republic','Croatia','Romania','Serbia','Austria',
    'Peru','Ecuador','Venezuela','Paraguay','Bolivia',
    'Senegal','Ivory-Coast','Tunisia','Mali','Burkina-Faso',
    'Saudi-Arabia','Qatar','United-Arab-Emirates','Kuwait','Uzbekistan','Costa-Rica'
]
developing_football = [
    'Lithuania','Latvia','Estonia','Malta','Bulgaria','Albania','Armenia','Georgia','Azerbaidjan','Kazakhstan',
    'Surinam','Nicaragua','Benin','Tanzania','Uganda','Zimbabwe',
    'Mongolia','Cambodia','Laos','Myanmar','Thailand','Lebanon','Malaysia','India','Vietnam','Philippines','Kyrgyzstan',
    'China','Iraq','Jordan','Oman','Turkmenistan','Panama','Belize','Barbados','Dominican-Republic','Cuba','Guatemala','Honduras',
    'Fiji','New-Zealand'
]
limited_development_football = [
    'San-Marino','Gibraltar','Cyprus','Faroe-Islands','Andorra','Iceland','Kosovo','Liechtenstein','Luxembourg','Macedonia',
    'Moldova','Montenegro','Northern-Ireland','Slovakia','Mauritania','Eswatini','Congo','Somalia','Malawi','Botswana','Angola',
    'Congo-DR','Sudan','Ethiopia','Guinea','Rwanda','Zambia','Burundi','Kenya','Bhutan','Nepal','Pakistan','Maldives',
    'Chinese-Taipei','Macao','Bermuda','Curacao','Guadeloupe','Trinidad-And-Tobago','Antigua-And-Barbuda','Jamaica','Haiti'
]

df['homeCountry_norm'] = df['homeTeamCountry'].apply(norm_country)

group_map = {}
for name in world_class_football:          group_map[norm_country(name)] = 'World class'
for name in strong_developing_leagues:     group_map[norm_country(name)] = 'Strong developing'
for name in developing_football:           group_map[norm_country(name)] = 'Developing'
for name in limited_development_football:  group_map[norm_country(name)] = 'Limited dev'

df['country_group'] = df['homeCountry_norm'].map(group_map).fillna('Other')
for col in ['Match Winner_Home','Match Winner_Draw','Match Winner_Away']:
    df[col] = pd.to_numeric(df[col], errors='coerce')

df['implied_home_prob'] = 1 / df['Match Winner_Home'].replace(0, np.nan)
df['implied_draw_prob'] = 1 / df['Match Winner_Draw'].replace(0, np.nan)
df['implied_away_prob'] = 1 / df['Match Winner_Away'].replace(0, np.nan)

df['is_home_win'] = (df['homeResult'] > df['awayResult']).astype(float)
df['is_draw']     = (df['homeResult'] == df['awayResult']).astype(float)
df['is_away_win'] = (df['homeResult'] < df['awayResult']).astype(float)

group_summary = (
    df.groupby('country_group')
      .agg(
          mean_implied_home=('implied_home_prob','mean'),
          mean_implied_draw=('implied_draw_prob','mean'),
          mean_implied_away=('implied_away_prob','mean'),
          actual_home=('is_home_win','mean'),
          actual_draw=('is_draw','mean'),
          actual_away=('is_away_win','mean'),
          n=('is_home_win','size')
      )
)

group_summary['diff_home'] = group_summary['actual_home'] - group_summary['mean_implied_home']
group_summary['diff_draw'] = group_summary['actual_draw'] - group_summary['mean_implied_draw']
group_summary['diff_away'] = group_summary['actual_away'] - group_summary['mean_implied_away']
group_summary = group_summary[group_summary['n'] >= 30]

print(group_summary[['n','diff_home','diff_draw','diff_away']].sort_values('n', ascending=False))
```

```python
def plot_diff_group(summary_df, diff_col, title):
    s = summary_df[diff_col].copy()
    s = s.reindex(s.abs().sort_values(ascending=False).index)
    plt.figure(figsize=(10,5))
    sns.barplot(x=s.index, y=s.values)
    plt.axhline(0, color='black', linestyle='--', linewidth=1)
    plt.title(title)
    plt.xlabel('Группа стран')
    plt.ylabel('Разница (actual - implied)')
    plt.xticks(rotation=15)
    plt.show()

plot_diff_group(group_summary, 'diff_home', 'Разница (actual - implied): Победа хозяев по группам стран')
plot_diff_group(group_summary, 'diff_draw', 'Разница (actual - implied): Ничья по группам стран')
plot_diff_group(group_summary, 'diff_away', 'Разница (actual - implied): Победа гостей по группам стран')
```
